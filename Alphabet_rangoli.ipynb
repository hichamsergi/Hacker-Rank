The center of the rangoli has the first alphabet letter a, and the boundary has the *N^th* alphabet letter (in alphabetical order).

#### Function Description

Complete the rangoli function in the editor below.

rangoli has the following parameters:

* *int size*: the size of the rangoli

#### Returns

* *string*: a single string made up of each of the lines of the rangoli separated by a newline character (\n).

````python
def print_rangoli(size):
    # your code goes here

    alph = ['a','b','c','d','e','f','g','h','i','j',
            'k','l','m','n','o','p','q','r','s','t',
            'u','v','w','x','y','z']
    
    r_alph = [alph[i] for i in range(size-1, -1, -1)]

    n_r_alph = [alph[i] for i in range(0, size)]

    #for r_num in range(size-1, -1, -1):

    #    print()

    print(r_alph, n_r_alph)



    

if __name__ == '__main__':
    n = int(input())
    print_rangoli(n)
````

````python
size = int(input())


alph = ['a','b','c','d','e','f','g','h','i','j',
            'k','l','m','n','o','p','q','r','s','t',
            'u','v','w','x','y','z']

r_alph = [alph[i] for i in range(size-1, -1, -1)]
n_alph = list(reversed(r_alph))[1:]

total_len = (size * 2) - 1

total_wid = (total_len * 2) - 1

for i in range(total_len//2):

    print(r_alph[i].center(total_wid, "-"))

for i in range(total_len//2, -1, -1):

    print(r_alph[i].center(total_wid, "-"))

#    r_alph = [alph[i] for i in range(size-1, -1, -1)]

#    n_alph = [alph[i] for i in range(1, size)]


print(r_alph+n_alph)
print(n_alph)
````
