### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Create 3 differents Series, each of length 100, as follows: 
1. The first a random number from 1 to 4 
2. The second a random number from 1 to 3
3. The third a random number from 10,000 to 30,000

```python
#1.
data1 = pd.Series(np.random.randint(1,5, size=100), name="1to4")

data2 = pd.Series(np.random.randint(1,4, size=100), name="1to3")

data3 = pd.Series(np.random.randint(10000, 30001, size=100), name="10kto30k")
```

### Step 3. Let's create a DataFrame by joinning the Series by column

```python
totals = pd.concat([data1, data2, data3], axis=1)

totals.head()
```

### Step 4. Change the name of the columns to bedrs, bathrs, price_sqr_meter

```python
house_info = totals.rename(columns={'1to4':'bedrs', '1to3':'bathrs', '10kto30k':'price_sqr_meter'})

house_info.head()
```

### Step 5. Create a one column DataFrame with the values of the 3 Series and assign it to 'bigcolumn'

```python
bigcolumn = pd.concat([data1, data2, data3], axis=0, ignore_index=True).to_frame()

bigcolumn.size
```
